## 3. Config

Switch `TRAIN_STEPS` from `50` to `2000` for full training.


```python
from pathlib import Path
TRAIN_STEPS = 50   # smoke: 50; full: 2000
SAVE_EVERY = 25 if TRAIN_STEPS <= 100 else 250
SAMPLE_EVERY = 25 if TRAIN_STEPS <= 100 else 250
PROJECT_NAME = 'mel_zit_smoke_a100' if TRAIN_STEPS <= 100 else 'mel_zit_full_a100'

yaml = f"""---
job: extension
config:
  name: "{PROJECT_NAME}"
  process:
    - type: 'sd_trainer'
      training_folder: "/content/zit_output"
      device: cuda:0
      trigger_word: "mel"
      network:
        type: "lora"
        linear: 16
        linear_alpha: 16
      save:
        dtype: float16
        save_every: {SAVE_EVERY}
        max_step_saves_to_keep: 8
      datasets:
        - folder_path: "/content/mel_zit_dataset"
          caption_ext: "txt"
          caption_dropout_rate: 0.05
          shuffle_tokens: false
          cache_latents_to_disk: true
          resolution: [512, 768]
      train:
        batch_size: 1
        steps: {TRAIN_STEPS}
        gradient_accumulation_steps: 1
        train_unet: true
        train_text_encoder: false
        gradient_checkpointing: true
        noise_scheduler: "flowmatch"
        optimizer: "adamw8bit"
        lr: 1e-4
        dtype: bf16
        ema_config:
          use_ema: true
          ema_decay: 0.99
      model:
        name_or_path: "Tongyi-MAI/Z-Image-Turbo"
        arch: "zimage"
        assistant_lora_path: "ostris/zimage_turbo_training_adapter/zimage_turbo_training_adapter_v1.safetensors"
        quantize: true
        qtype: "qfloat8"
        quantize_te: true
        qtype_te: "qfloat8"
        low_vram: true
      sample:
        sampler: "flowmatch"
        sample_every: {SAMPLE_EVERY}
        width: 768
        height: 768
        prompts:
          - "mel, 1girl, silver white hair, purple eyes, white cat ears, white cat tail, cute anime catgirl, school uniform, clean lineart, high quality"
          - "mel, 1girl, silver white hair, purple eyes, white cat ears, white cat tail, casual outfit, standing, soft smile, high quality"
        neg: "bad hands, extra fingers, deformed fingers, blurry, low quality"
        seed: 42
        walk_seed: true
        guidance_scale: 0.0
        sample_steps: 8
meta:
  name: "{PROJECT_NAME}"
  version: "1.0"
"""
Path('/content/mel_zit_train.yaml').write_text(yaml, encoding='utf-8-sig')
print(yaml)
```
